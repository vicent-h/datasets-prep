```python
import pandas as pd
import pyarrow.parquet as pq
```

```python
df_analise = pd.read_parquet('out/analise_textos_inicial.parquet')
```

```python
line_to_show = df_analise[
    df_analise['eval'] == False
].tail(20).reset_index().iloc[3, :]
with (open(line_to_show['path'] + line_to_show['src_file'], 'r', encoding='utf-8') as f_src, 
    open(line_to_show['path'] + line_to_show['tgt_file'], 'r', encoding='utf-8') as f_tgt):
    for i, (src_line, tgt_line) in enumerate(zip(f_src, f_tgt)):
        if (i == line_to_show['index']):
            print(src_line, tgt_line)
            break
```
